This is a simple Hangman game using Python programming language. You can use this as a project to boost their programming skills and understanding logic.  
The Hangman program randomly selects a secret word from a list of secret words. The random module will provide this ability, so line 1 in program imports it.
The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to win the game.
When a letter in that word is guessed correctly, that letter position in the word is made visible. In this way, all letters of the word are to be guessed before all the chances are over. 
For convenience, we have given length of word + 2 chances. For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five-letter word.



```python
import random as rd
```

```python
list_of_fruits = ["apple", "mango", "grapes", "pineapple", "pomegranate"]

# choice of word to print from list
word = rd.choice(list_of_fruits)

# to display the underscore for further updation
display = ['_'] * len(word)

# create variable for getting chances.
chances = len(word) + 2

# Create a flag variable to track user quits or not.
game_stopped = False
# A flag variable is a variable used to keep track of whether a particular event or condition has occurred in a program.

print("===== HANGMAN GAME =====")
print("Guess the fruit name!")

# Repeat loop until chances become 0
while chances > 0:

    # Display current word status here
    print("\nWord :", " ".join(display))
    print("Remaining Chances :", chances)

    # Take input from user
    letter = input("Enter a letter : ").lower()
    if letter in word:
        # Check each character in word
        for i in range(len(word)):
            if word[i] == letter:
                display[i] = letter
        print("Correct Guess!")
        # If letter not found, reduce chance

    elif letter == "quit":
        print("\nGame Stopped by User!")
        print("The secret word was:", word)
        game_stopped = True
        break
        
    else:
        chances -= 1
        print("Wrong Guess!")
        
    # Check if word is completely guessed
    if "_" not in display:
        print("\n Congratulations!")
        print("You guessed the word :", word)
        break

# If chances become 0
if not game_stopped and "_" in display:
    print("\nGame Over!")
    print("Correct word was :", word)
```
